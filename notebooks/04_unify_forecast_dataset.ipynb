# Imports

```python
import pandas as pd

from pathlib import Path
```

# Fixed variables

```python
PROJECT_ROOT = Path.cwd().parent

DATA_DIR = PROJECT_ROOT / "data"

INPUT_DATA_DIR = DATA_DIR / "input"

redes_path = INPUT_DATA_DIR / "01_redes.csv"
meteo_path = INPUT_DATA_DIR / "02_meteo.csv"
holidays_path = INPUT_DATA_DIR / "03_holidays.csv"

output_path = INPUT_DATA_DIR / "04_input.csv"
```

# Get Data

```python
df_redes = pd.read_csv(redes_path)
df_meteo = pd.read_csv(meteo_path)
df_holidays = pd.read_csv(holidays_path)
```

```python
df_redes['datetime'] = pd.to_datetime(df_redes['datetime'], utc=True).dt.normalize()
df_meteo['fecha'] = pd.to_datetime(df_meteo['fecha'], utc=True).dt.normalize()
df_holidays['date'] = pd.to_datetime(df_holidays['date'], utc=True).dt.normalize()
```

```python
len(df_redes), len(df_meteo), len(df_holidays)
```

# Process Data

```python
df_total_1 = df_redes.merge(
    df_meteo, how="inner", left_on="datetime", right_on="fecha"
).drop(columns=["datetime", "percentage"])
```

```python
df_total_2 = df_total_1.merge(
    df_holidays, how="left", left_on="fecha", right_on="date"
).drop(columns=["date"])
```

```python
df_total_2["is_holiday"] = df_total_2["holiday_name"].isna() == False
```

```python
df_total = df_total_2.rename(columns={"value": "demand"}). \
    drop(columns=["holiday_name"])[["fecha", "demand", "tmed", "tmin", "tmax", "prec", "velmedia", "is_holiday"]]
```

# Save Data

```python
df_total.to_csv(output_path, index=False)

print(f"Dataset saved to:\n{output_path}")
```
