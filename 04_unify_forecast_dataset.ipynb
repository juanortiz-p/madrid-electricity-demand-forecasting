# Imports

```python
import pandas as pd

from pathlib import Path
```

```python
PROJECT_ROOT = Path.cwd().parent

DATA_DIR = PROJECT_ROOT / "data"

INPUT_DATA_DIR = DATA_DIR / "input"

redes_path = INPUT_DATA_DIR / "01_redes.csv"
meteo_path = INPUT_DATA_DIR / "02_meteo.csv"
holidays_path = INPUT_DATA_DIR / "03_holidays.csv"
```

```python
df_redes = pd.read_csv(redes_path)
df_meteo = pd.read_csv(meteo_path)
df_holidays = pd.read_csv(holidays_path)
```

```python
df_redes['datetime'] = pd.to_datetime(df_redes['datetime'], utc=True)
df_meteo['fecha'] = pd.to_datetime(df_meteo['fecha'], utc=True)
df_holidays['date'] = pd.to_datetime(df_holidays['date'], utc=True)
```
